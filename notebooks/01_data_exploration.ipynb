## Data Quality Assessment
Before proceeding, we need to identify the extent of missing data. Columns with excessive missing values may not be useful for prediction and might need to be dropped.

### Missing Values

```python
# Calculate percentage of missing values per column
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Missing_Percentage': missing_percentage})

# Display columns with more than 50% missing data
print(missing_df[missing_df['Missing_Percentage'] > 50])
```

### Data Integrity Check: Duplicates
It is essential to check for duplicate entries in the dataset, as they can bias the machine learning model and lead to overfitting.

```python
# Check for duplicate rows
duplicates_count = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicates_count}")
```
